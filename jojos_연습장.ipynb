```python
#영수증 형태로 출력되는 Coffee 클래스
class Coffee:
    def __init__(self, name, size, temperature, price):
        self.name = name
        self.size = size
        self.temperature = temperature
        self.price = price

    def receipt(self):
        print("====== 영수증 ======")
        print("메뉴:", self.name)
        print("사이즈:", self.size)
        print("온도:", self.temperature)
        print("가격:", self.price, "원")
        print("-------------------")
        print("이용해 주셔서 감사합니다 😊")

#객체생성 (영수증)
coffee1 = Coffee("아메리카노", "M", "hot", 4500)
coffee2 = Coffee("라떼", "L", "cold", 5500)

#영수증 출력
coffee1.receipt()
coffee2.receipt()
```

```python
# TodayToDo 클래스
# 오늘 영화관에서 할 일을 정리한 투두리스트 클래스
class TodayToDo:

    # __init__ 메서드
    # 객체가 만들어질 때 자동으로 실행되며,
    # 오늘 볼 영화, 먹을 팝콘, 마실 커피, 현재 시간을 저장한다.
    def __init__(self, movie, popcorn, coffee, time_hour):
        self.movie = movie            # 영화 제목
        self.popcorn = popcorn        # 팝콘 종류
        self.coffee = coffee          # 커피 종류
        self.time_hour = time_hour    # 현재 시간 (숫자)
        self.dinner_list = ["파스타", "김치찌개"]  # 저녁 메뉴 리스트

    # 제목을 출력하는 메서드
    def show_title(self):
        print("====== TODAY TO DO LIST ======")

    # 영화 보기 행동을 출력하는 메서드
    def watch_movie(self):
        print("1. 영화 보기:", self.movie)

    # 팝콘 먹기 행동을 출력하는 메서드
    def eat_popcorn(self):
        print("2. 팝콘 먹기:", self.popcorn)

    # 커피 마시기 행동을 출력하는 메서드
    def drink_coffee(self):
        print("3. 커피 마시기:", self.coffee)

    # 저녁 먹기 여부를 출력하는 메서드
    # 시간이 6시(18시) 이후면 저녁 메뉴를 출력한다.
    def eat_dinner(self):
        if self.time_hour >= 18:
            print("4. 저녁 먹기:", self.dinner_list[0], ",", self.dinner_list[1])
        else:
            print("4. 저녁 먹기: 아직 저녁 시간이 아닙니다.")

    # 마지막 인사 멘트를 출력하는 메서드
    def finish_message(self):
        print("-----------------------------")
        print("오늘도 수고했어요. 즐거운 하루 보내세요 😊")


# -------------------------------
# 객체 생성 (오늘의 투두리스트 만들기)
# -------------------------------
today = TodayToDo("듄 2", "카라멜 팝콘", "아이스 아메리카노", 19)

# -------------------------------
# 메서드 호출 (출력 실행)
# -------------------------------
today.show_title()      # 제목 출력
today.watch_movie()     # 영화 보기 출력
today.eat_popcorn()     # 팝콘 먹기 출력
today.drink_coffee()    # 커피 마시기 출력
today.eat_dinner()      # 저녁 먹기 출력
today.finish_message()  # 마무리 멘트 출력
```

```python
class Account:
    # 생성자: 예금주와 초기 잔액만 입력받음
    def __init__(self, owner, balance):
        self.bank = "SC은행"        # 은행 이름
        self.owner = owner          # 예금주
        self.balance = balance      # 잔액
        self.account_number = ""    # 계좌번호 (처음엔 없음)

    # 계좌 개설 메서드
    def open_account(self):
        # 계좌번호 형식: 3자리-2자리-6자리
        # (랜덤 대신 형식만 맞춰 생성)
        self.account_number = "111-11-111111"

    # 계좌 정보 출력 메서드
    def show_info(self):
        print("은행명:", self.bank)
        print("예금주:", self.owner)
        print("계좌번호:", self.account_number)
        print("잔액:", self.balance, "원")
account1 = Account("홍길동", 100000)
account1.open_account()
account1.show_info()
```
