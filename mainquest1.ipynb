```python
class Account: #메인 퀘스트 1번: 은행계좌 만들기
    # 생성자: 예금주와 초기 잔액만 입력받음
    def __init__(self, owner, balance):
        self.bank = "SC은행"        # 은행 이름
        self.owner = owner          # 예금주
        self.balance = balance      # 잔액
        self.account_number = ""    # 계좌번호 (처음엔 없음)

    # 계좌 개설 메서드
    def open_account(self):
        # 계좌번호 형식: 3자리-2자리-6자리
        # (랜덤 대신 형식만 맞춰 생성)
        self.account_number = "111-11-111111"

    # 계좌 정보 출력 메서드
    def show_info(self):
        print("은행명:", self.bank)
        print("예금주:", self.owner)
        print("계좌번호:", self.account_number)
        print("잔액:", self.balance, "원")
account1 = Account("조수경", 100000)
account1.open_account()
account1.show_info()
```
