```python
class Account:
    account_count = 0   # 클래스 변수 (전체 계좌 수)

    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = balance
        Account.account_count += 1  # 계좌 하나 생성될 때마다 +1
```

```python
a = Account("수경", 10000)
b = Account("민수", 5000)

print(Account.account_count)
```
